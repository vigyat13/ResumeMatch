```python
def flatten_tags(data, parent_path=''):
    flattened = []
    
    for item in data:
        current_path = f"{parent_path}/{item['name']}" if parent_path else item['name']
        flat_item = {k: v for k, v in item.items() if k != 'tags'}
        flat_item['full_path'] = current_path
        flattened.append(flat_item)
        
        if 'tags' in item and isinstance(item['tags'], list):
            flattened.extend(flatten_tags(item['tags'], current_path))
    
    return flattened
```

```python
all_flattened = []

for row in json['tags']:
    if isinstance(row, list):
        flattened = flatten_tags(row)
    else:
        flattened = flatten_tags([row])
    all_flattened.extend(flattened)

flat_df = pd.DataFrame(all_flattened)
```
